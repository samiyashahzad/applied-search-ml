# Research Question and Provisional Lane

## My Lane

**Lane:** CTR / Engagement Opportunity Scoring

### Why I chose this lane

SEO teams often manage thousands of pages but have time to review only a small number each week. Rather than trying to predict whether a page is simply "good" or "bad," I want to build a ranking system that prioritizes pages with the greatest opportunity for improvement.

The goal is to estimate which pages deserve review first by considering multiple observable signals such as impressions, click-through rate (CTR), search position, and engagement metrics. The output is intended to support human decision-making rather than replace it.


## Research Question

**Which pages should an SEO team review first because they appear to underperform in clicks or engagement relative to their visibility?**

### Decision

Given limited time, which pages should the SEO team review this week?

### Action

The reviewer investigates the highest-ranked pages and decides whether to:
- improve titles or meta descriptions
- refresh content
- improve content quality
- improve user engagement
- monitor the page instead of changing it

### Output

Instead of predicting a simple yes/no label, the system produces an **Opportunity Score** and a **ranked list of pages** with reason codes explaining why each page received its score.

### Cost of Wrong Recommendation

**False Positive**

The model ranks a page very highly even though it isn't actually worth reviewing.
Cost: The SEO team wastes time.

**False Negative**

A valuable page with genuine opportunity isn't recommended.
Cost: The business may miss additional traffic or engagement because the page was never reviewed.

### Why ML?

The relationship between impressions, CTR, position, freshness, engagement, and content characteristics is more complex than a single rule. A machine learning ranking model may identify useful combinations of signals that help prioritize review candidates more effectively than a fixed scoring rule. The goal is to improve prioritization rather than automate SEO decisions.


## Quick Look at the Data

Let's see why this lane is worth our time by looking at a few numbers from the dataset.


```python
import pandas as pd
from pathlib import Path

csv_path = Path('../../data/raw/content_refresh_anonymized.csv')
df = pd.read_csv(csv_path)

page_count = len(df)
avg_ctr = df['ctr'].mean()
avg_engagement = df['engagement_rate'].mean()

# Pages that are visible but have a CTR below 1%
high_impression_low_ctr = int(((df['impressions_90d'] > 500) & (df['ctr'] < 1.0)).sum())
very_visible_pages = int((df['impressions_90d'] > 1000).sum())

print(f'Total pages in dataset: {page_count:,}')
print(f'Average CTR: {avg_ctr:.2f}%')
print(f'Average engagement rate: {avg_engagement:.2f}%')
print(f'Pages with >1000 impressions: {very_visible_pages:,} ({(very_visible_pages/page_count)*100:.1f}%)')
print(f'Pages with >500 impressions but CTR < 1%: {high_impression_low_ctr:,}')
```
