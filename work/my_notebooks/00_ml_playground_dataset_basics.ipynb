## Rule #1

Not every column should become a feature.

A good feature gives useful information about the target.

An ID usually doesn't help because it only identifies a record.

A leaked column is even worse because it already contains the answer we are trying to predict.

```python
[col for col in ["trend_direction", "trend_pct", "is_declining_label"] if col in df.columns]
```
